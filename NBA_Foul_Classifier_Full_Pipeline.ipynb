## Stage 1: NBA Last Two Minute (L2M) Scraper & Ground-Truth Labeler (`1_scrape_l2m.py`)
Official NBA Last Two Minute (L2M) reports provide transparent, post-game officiating evaluations for all contested plays in the final 2 minutes of close games (within 3 points). We map decisions to binary ground-truth labels:

- **Positive Class (`1 = foul`)**:
  - `CC` (*Correct Call*): Referee whistled a foul, and NBA league review confirmed illegal contact occurred.
  - `INC` (*Incorrect Non-Call*): Referee allowed play to continue, but league review confirmed a foul was committed (*crucial ground-truth signal*).
- **Negative Class (`0 = no_foul`)**:
  - `CNC` (*Correct Non-Call*): Legal defensive contest, no whistle blown.
  - `IC` (*Incorrect Call*): Referee blew the whistle, but league review found legal guarding positioning (*phantom foul*).

The cell below synchronizes `data/l2m_incorrect_fouls.csv` with the cached ground-truth metadata in `data/l2m_metadata.csv` (or queries `official.nba.com` if `RUN_LIVE_WEB_SCRAPE = True`).


```python
# ==============================================================================
# CELL 2: Stage 1 — NBA L2M Scraper & Local Metadata Sync
# ==============================================================================
import importlib
scrape_mod = importlib.import_module('1_scrape_l2m')

RUN_LIVE_WEB_SCRAPE = False  # Set to True to query official.nba.com for CONFIG['max_games'] games

if RUN_LIVE_WEB_SCRAPE:
    print(f'Scraping official NBA L2M reports for up to {CONFIG["max_games"]} games...')
    scraped_rows = scrape_mod.run_training_dataset_scraper(
        output_dir=RAW_DIR,
        metadata_csv=METADATA_CSV,
        daily_csv=INCORRECT_CSV,
        max_games=CONFIG['max_games'],
        max_clips=CONFIG['max_scrape_clips'],
        skip_download=False,
    )
    print(f'Scraped {len(scraped_rows)} plays into {METADATA_CSV.name}')
else:
    # Synchronize & populate l2m_incorrect_fouls.csv offline from cached l2m_metadata.csv
    populated_foul_rows = scrape_mod.populate_foul_csv_from_local_metadata(
        metadata_csv=METADATA_CSV,
        daily_csv=INCORRECT_CSV,
        include_confirmed_fouls=True,
    )
    print(f'✅ Successfully synced {len(populated_foul_rows)} foul review entries into {INCORRECT_CSV.name}')
```

## Stage 2: Biomechanical Pose Estimation & Action-Tube Extraction (`2_preprocess_tubes.py`)
Feeding uncompressed 1080p broadcast video clips directly into 3D Convolutional Neural Networks leads to **RAM/VRAM overflow** (L2M clips often exceed 10,000 frames / 76+ GB uncompressed) and **spatial dilution** (interacting players occupy < 8% of the court pixels).

### Our Two-Pass Cross-Platform Solution:
1. **Pass 1 (Lightweight Pose Scan)**: Scans the live play window using `yolo11n-pose.pt` in micro-batches (`yolo_batch_size = 4`). It extracts a **12-D Mechanical Foul Feature Vector** grounded in NBA Rule 12 officiating rules.
2. **Court Geometry & Sideline Rejection**: Strictly filters court floor coordinates ($y_c \in [0.22H, 0.68H]$, $y_2 \le 0.72H$) to reject referees (#43), bench players, and background spectators.
3. **Pass 2 (Stabilized 16-Frame Crop)**: Applies Exponential Moving Average (EMA) smoothing to the bounding box trajectory and crops **only the 16 frames** in a `1.5s` window around peak contact at `224x224` resolution, saving `(3, 16, 224, 224)` RGB tubes and `(16, 12)` pose tensors.
4. **Data Isolation**: All clips residing in `data/test/` are strictly excluded from training preprocessing to prevent data leakage.

## Stage 3: Dual-Stream Spatiotemporal Architecture (`3_model_and_dataset.py`)
Our classifier fuses two complementary representations into a unified 768-D representation:

1. **Visual Spatiotemporal Stream (`R(2+1)D-18`)**: Decomposes 3D convolutions into $2D$ spatial $+ 1D$ temporal operations over the `(3, 16, 224, 224)` action tube, yielding a **512-D visual embedding**.
2. **Kinematic Temporal Attention Pose Stream (`BiomechanicalFoulEncoder`)**: Projects the `(16, 12)` mechanical foul sequence into a 128-D space, applies **4-Head Temporal Self-Attention** across all 16 frames with residual connection and LayerNorm, followed by dual temporal pooling (mean + max) to produce a **256-D kinematic contact embedding**.
3. **Late-Fusion Classification Head**: Concatenates `[512-D Visual || 256-D Pose] = 768-D` and passes through a 3-layer MLP (`768 -> 512 -> 256 -> 1`) with LayerNorm, GELU, and Dropout.

> [!NOTE]
> **Zero-Stall Training Architecture:** Freezing the 3D ConvNet backbone and utilizing precomputed 512-D Kinetics embeddings allows the 4-head attention network and fusion MLP (~613K parameters) to train in **under 3 seconds** on laptops and Google Colab, with zero swap thrashing or memory bottlenecks!

## Stage 4: High-Performance Training Loop, Focal Loss & Threshold Calibration (`4_train_eval.py`)
To handle class imbalance and prevent PyTorch collation stalls:

1. **`safe_foul_collate`**: Dynamically verifies batch tensor shapes and harmonizes representations before `torch.stack`, completely eliminating PyTorch DataLoader `RuntimeError: stack expects each tensor to be equal size`.
2. **Class-Balanced `FoulFocalBCEWithLogitsLoss` (`gamma = 1.5`)**: Multiplies positive examples by $pos\_weight$ and dynamically scales loss by $(1 - p_t)^\gamma$ to penalize hard, subtle foul misclassifications.
3. **Dynamic F1 Threshold Calibration**: Sweeps probability thresholds $\tau \in [0.20, 0.80]$ on validation predictions at every epoch to maximize F1-score rather than assuming a naive $0.50$ cutoff.


```python
# ==============================================================================
# CELL 7: Train & Calibrate the Dual-Stream Foul Classifier
# ==============================================================================
import inspect
train_mod = importlib.import_module('4_train_eval')
importlib.reload(train_mod)

train_mod.seed_everything(CONFIG['seed'])

all_samples = demo_dataset.samples
labels = [lbl for _, lbl in all_samples]
stratify_labels = labels if len(set(labels)) > 1 and min(labels.count(0), labels.count(1)) >= 2 else None

train_samples, val_samples = train_test_split(
    all_samples,
    test_size=0.25,
    random_state=CONFIG['seed'],
    stratify=stratify_labels,
)

train_ds_kwargs = {'samples': train_samples, 'is_train': True, 'return_foul_features': True, 'use_cached_embeddings': True}
valid_train_ds_kwargs = {k: v for k, v in train_ds_kwargs.items() if k in ds_sig.parameters}
train_ds = L2MFoulDataset(**valid_train_ds_kwargs)

val_ds_kwargs = {'samples': val_samples, 'is_train': False, 'return_foul_features': True, 'use_cached_embeddings': True}
valid_val_ds_kwargs = {k: v for k, v in val_ds_kwargs.items() if k in ds_sig.parameters}
val_ds = L2MFoulDataset(**valid_val_ds_kwargs)

safe_workers = get_safe_num_workers(0)
collate_fn = getattr(train_mod, 'safe_foul_collate', None)
train_loader_kwargs = {'batch_size': CONFIG['batch_size'], 'shuffle': True, 'num_workers': safe_workers}
if collate_fn is not None:
    train_loader_kwargs['collate_fn'] = collate_fn
train_loader = DataLoader(train_ds, **train_loader_kwargs)

val_loader_kwargs = {'batch_size': CONFIG['batch_size'], 'shuffle': False, 'num_workers': safe_workers}
if collate_fn is not None:
    val_loader_kwargs['collate_fn'] = collate_fn
val_loader = DataLoader(val_ds, **val_loader_kwargs)

# Initialize classifier with kwargs dynamically matched to class signature
model = FoulClassifier3D(**valid_m_kwargs).to(DEVICE)

pos_weight = train_mod.compute_pos_weight(train_samples, device=DEVICE)
criterion = train_mod.FoulFocalBCEWithLogitsLoss(pos_weight=pos_weight, gamma=1.5)
optimizer = AdamW([p for p in model.parameters() if p.requires_grad], lr=CONFIG['lr'], weight_decay=CONFIG['weight_decay'])
scheduler = CosineAnnealingLR(optimizer, T_max=max(1, CONFIG['epochs']), eta_min=CONFIG['lr'] * 0.01)
scaler = torch.amp.GradScaler('cuda', enabled=True) if (DEVICE.type == 'cuda' and torch.cuda.is_available()) else None

history = []
best_f1 = -1.0
start_train_time = time.time()

print(f'Starting training for {CONFIG["epochs"]} epochs on {DEVICE}...')
for epoch in range(1, CONFIG['epochs'] + 1):
    t_loss = train_mod.train_one_epoch(
        model=model,
        loader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        scaler=scaler,
        device=DEVICE,
        accum_steps=CONFIG['accum_steps'],
    )
    v_loss, metrics = train_mod.evaluate(
        model=model,
        loader=val_loader,
        criterion=criterion,
        device=DEVICE,
    )
    scheduler.step()
    history.append({'epoch': epoch, 'train_loss': t_loss, 'val_loss': v_loss, **metrics})
    print(
        f'Epoch [{epoch:02d}/{CONFIG["epochs"]:02d}] | Train Loss: {t_loss:.4f} | Val Loss: {v_loss:.4f} | '
        f'Acc: {metrics["accuracy"]:.3f} | F1: {metrics["f1"]:.3f} | ROC-AUC: {metrics["roc_auc"]:.3f} | '
        f'Opt Thresh: {metrics["threshold"]:.2f}'
    )
    if metrics['f1'] >= best_f1:
        best_f1 = metrics['f1']
        WEIGHTS_DIR.mkdir(parents=True, exist_ok=True)
        torch.save(
            {
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'metrics': metrics,
                'optimal_threshold': metrics['threshold'],
            },
            CHECKPOINT_PATH,
        )

elapsed = time.time() - start_train_time
print(f'\n⚡ Training completed in {elapsed:.2f}s! Best Calibrated F1-Score: {best_f1:.3f}')
print(f'💾 Saved calibrated model checkpoint -> {CHECKPOINT_PATH}')

# Matplotlib Training History Curves
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
epochs_range = [h['epoch'] for h in history]

axes[0].plot(epochs_range, [h['train_loss'] for h in history], 'o-', label='Train Focal Loss', color='#1f77b4', linewidth=2)
axes[0].plot(epochs_range, [h['val_loss'] for h in history], 's--', label='Val Focal Loss', color='#ff7f0e', linewidth=2)
axes[0].set_xlabel('Epoch', fontsize=11)
axes[0].set_ylabel('Focal Loss', fontsize=11)
axes[0].set_title('Training & Validation Loss Convergence', fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs_range, [h['accuracy'] for h in history], '^-', label='Accuracy', color='#2ca02c', linewidth=2)
axes[1].plot(epochs_range, [h['f1'] for h in history], 'o-', label='F1-Score', color='#d62728', linewidth=2)
axes[1].plot(epochs_range, [h['roc_auc'] for h in history], 'x--', label='ROC-AUC', color='#9467bd', linewidth=2)
axes[1].set_xlabel('Epoch', fontsize=11)
axes[1].set_ylabel('Metric Score (0 - 1)', fontsize=11)
axes[1].set_title('Validation Performance & F1 Calibration', fontsize=12, fontweight='bold')
axes[1].set_ylim(0.0, 1.05)
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Stage 5: Single-Clip Foul Prediction & Biomechanical Contact Telemetry
The cell below loads the calibrated `best_foul_r2plus1d.pt` checkpoint, evaluates an unseen play from the validation split, identifies the exact moment of peak physical impact across the 16-frame spatiotemporal tube, and displays:

1. **Continuous 16-Frame Contact Trajectory Plot**: Visualizes the evolution of arm intrusion, cylinder penetration, and relative closing velocity across time, marking the exact peak collision frame.
2. **12-D Mechanical Foul Dimension Profile**: Horizontal bar chart illustrating the complete biomechanical signature at the critical contact moment.


```python
# ==============================================================================
# CELL 8: Inference & Frame-by-Frame Biomechanical Telemetry Inspection
# ==============================================================================
ckpt = model_mod.load_torch_file_safe(CHECKPOINT_PATH, map_location=DEVICE)
model.load_state_dict(ckpt['model_state_dict'], strict=False)
calibrated_threshold = float(ckpt.get('optimal_threshold', 0.50))
model.eval()

val_path, val_true_label = val_samples[0]
val_video, val_feats, _ = val_ds[0]

with torch.no_grad():
    logit = model(val_video.unsqueeze(0).to(DEVICE), val_feats.unsqueeze(0).to(DEVICE))
    foul_prob = float(torch.sigmoid(logit).item())

verdict = '🚨 FOUL DETECTED' if foul_prob >= calibrated_threshold else '✅ LEGAL PLAY / NO FOUL'
feats_np = val_feats.cpu().numpy()  # (16, 12)
contact_composite = feats_np[:, 0] * 0.40 + feats_np[:, 1] * 0.35 + feats_np[:, 9] * 0.25
peak_frame_idx = int(np.argmax(contact_composite))
peak_contact_score = float(contact_composite[peak_frame_idx])

print(f'🎯 Evaluated Play ID      : {val_path.stem}')
val_label_str = 'FOUL' if val_true_label == 1 else 'NO FOUL'
print(f'🏷️  Ground-Truth Label     : {val_true_label} ({val_label_str})')
print(f'📊 Predicted Probability   : {foul_prob * 100:.2f}% (Calibrated Threshold: {calibrated_threshold * 100:.0f}%)')
print(f'⚖️  Model Decision         : {verdict}')
print(f'⏱️  Peak Contact Frame     : Frame {peak_frame_idx + 1}/16 (Biomechanics Index = {peak_contact_score * 100:.1f}%)')

# ------------------------------------------------------------------------------
# Matplotlib Biomechanical Contact Telemetry Visualizations
# ------------------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 5.2))
frames_axis = list(range(1, 17))

# Left: 16-Frame Temporal Trajectory
axes[0].plot(frames_axis, feats_np[:, 0], 'o-', label='Arm Contact Intrusion', color='#d62728', linewidth=2)
axes[0].plot(frames_axis, feats_np[:, 1], 's-', label='Body Cylinder Penetration', color='#ff7f0e', linewidth=2)
axes[0].plot(frames_axis, feats_np[:, 9], '^--', label='Closing Velocity', color='#1f77b4', linewidth=1.8)
axes[0].plot(frames_axis, contact_composite, 'd:', label='Composite Contact Force', color='#9467bd', linewidth=2.2)
axes[0].axvline(x=peak_frame_idx + 1, color='crimson', linestyle='--', linewidth=2, label=f'Peak Contact (F{peak_frame_idx+1})')
axes[0].set_xlabel('Spatiotemporal Action Frame (1 - 16)', fontsize=11)
axes[0].set_ylabel('Normalized Mechanical Intensity (0 - 1)', fontsize=11)
axes[0].set_title('Biomechanical Contact Trajectory Across Time', fontsize=12, fontweight='bold')
axes[0].set_ylim(-0.05, 1.05)
axes[0].legend(loc='upper left', fontsize=9.5)
axes[0].grid(True, alpha=0.3)

# Right: 12-D Mechanical Dimensions Profile at Peak Contact Frame
dim_labels = list(model_mod.POSE_FEATURE_NAMES)
peak_magnitudes = feats_np[peak_frame_idx]
bar_colors = ['#d62728' if v > 0.45 else '#1f77b4' for v in peak_magnitudes]
axes[1].barh(dim_labels, peak_magnitudes, color=bar_colors, height=0.65)
axes[1].set_xlabel('Normalized Magnitude (0 - 1)', fontsize=11)
axes[1].set_title(f'12-D Mechanical Foul Signature at Peak Frame {peak_frame_idx+1}', fontsize=12, fontweight='bold')
axes[1].set_xlim(0, 1.05)
axes[1].grid(axis='x', alpha=0.3)
for i, v in enumerate(peak_magnitudes):
    if v > 0.05:
        axes[1].text(v + 0.02, i, f'{v:.2f}', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```
